# Convert table to adjacency matrix

```python

import pandas as pd
import numpy as np

city_distances = pd.read_csv("../data/city_distances.csv")
```

## Drive distance Adjacency Matrix

```python
# Get unique cities
cities = pd.concat([city_distances['from'], city_distances['to']]).unique()

# Create an empty adjacency matrix
drive_distance_adj_mat = pd.DataFrame(np.inf, index=cities, columns=cities)

# Fill the adjacency matrix with distances
for _, row in city_distances.iterrows():
    drive_distance_adj_mat.at[row['from'], row['to']] = row['distance_km']
    drive_distance_adj_mat.at[row['to'], row['from']] = row['distance_km']  

# Replace inf with 0 for cities referencing themselves
drive_distance_adj_mat.replace(np.inf, 0, inplace=True)

display(drive_distance_adj_mat)
```

## Drive time Adjacency Matrix

```python
# Get unique cities
cities = pd.concat([city_distances['from'], city_distances['to']]).unique()

# Create an empty adjacency matrix
drive_time_adj_mat = pd.DataFrame(np.inf, index=cities, columns=cities)

# Fill the adjacency matrix with distances
for _, row in city_distances.iterrows():
    drive_time_adj_mat.at[row['from'], row['to']] = row['duration_min']
    drive_time_adj_mat.at[row['to'], row['from']] = row['duration_min']  

# Replace inf with 0 for cities referencing themselves
drive_time_adj_mat.replace(np.inf, 0, inplace=True)
 
display(drive_time_adj_mat)
```

## Save into new CSVs

```python
drive_distance_adj_mat.to_csv("../data/adjacencyMatrixDist.csv")
drive_time_adj_mat.to_csv("../data/adjacencyMatrixTravelTime.csv")
```
