# Schedule playground

## Helper functions

```python
import pandas as pd

def get_dow(d):

    m = d.to_pydatetime().weekday()

    if m == 0: 
        return "Monday"
    elif m == 1: 
        return "Tuesday"
    elif m == 2: 
        return "Wednesday"
    elif m == 3: 
        return "Thursday"
    elif m == 4: 
        return "Friday"
    elif m == 5: 
        return "Saturday"
    elif m == 6: 
        return "Sunday"

def get_month(d):
    m = d.to_pydatetime().month

    if m == 1: 
        return "January"
    elif m == 2: 
        return "February"
    elif m == 3: 
        return "March"
    elif m == 4: 
        return "April"
    elif m == 5: 
        return "May"
    elif m == 6: 
        return "June"
    elif m == 7: 
        return "July"
    elif m == 8: 
        return "August"
    elif m == 9: 
        return "September"
    elif m == 10: 
        return "October"
    elif m == 11: 
        return "November"
    elif m == 12: 
        return "December"

def create_empty_sched_df(start_date, end_date, num_rows):
    # Generate a date range from start_date to end_date
    date_range = pd.date_range(start=start_date, end=end_date)

    # Create a DataFrame with the date range as columns and num_rows as rows, initialized to False
    date_range = [f"{str(d)[0:10]}_{get_month(d)}_{get_dow(d)}" for d in date_range ]

    df = pd.DataFrame(False, index=range(num_rows), columns=date_range)

    return df

```

```python


start_date = '2025-08-25'  # Full start date
end_date = '2025-12-15'     # Full end date
num_rows = 5                # Number of rows

result_df = create_empty_sched_df(start_date, end_date, num_rows)
result_df
```
