# Schedule playground

## Helper functions

```python
import pandas as pd

def get_dow(d):

    m = d.to_pydatetime().weekday()

    if m == 0: 
        return "Monday"
    elif m == 1: 
        return "Tuesday"
    elif m == 2: 
        return "Wednesday"
    elif m == 3: 
        return "Thursday"
    elif m == 4: 
        return "Friday"
    elif m == 5: 
        return "Saturday"
    elif m == 6: 
        return "Sunday"

def get_month(d):
    m = d.to_pydatetime().month

    if m == 1: 
        return "January"
    elif m == 2: 
        return "February"
    elif m == 3: 
        return "March"
    elif m == 4: 
        return "April"
    elif m == 5: 
        return "May"
    elif m == 6: 
        return "June"
    elif m == 7: 
        return "July"
    elif m == 8: 
        return "August"
    elif m == 9: 
        return "September"
    elif m == 10: 
        return "October"
    elif m == 11: 
        return "November"
    elif m == 12: 
        return "December"

def create_empty_sched_df(start_date, end_date, num_rows, bHours=(8,20), bDays=list(range(0,5))):

    sod = bHours[0]
    eod = bHours[1]

    # Generate a date range from start_date to end_date
    date_range = pd.date_range(start=start_date, end=end_date, freq="30min")

    # Filter out non-business hours
    business_hours = date_range[(date_range.hour >= sod) & (date_range.hour < eod) ]

    business_days = business_hours[~business_hours.weekday.isin([5, 6])]

    # Create a DataFrame with the date range as columns and num_rows as rows, initialized to False
    business_days = [f"{str(d)}_{get_month(d)}_{get_dow(d)}" for d in business_days ]

    df = pd.DataFrame(False, index=range(num_rows), columns=business_days)

    return df

dow = {
    "Monday": 0,
    "Tuesday": 1,
    "Wednesday": 2,
    "Thursday": 3,
    "Friday": 4,
    "Saturday": 5,
    "Sunday": 6,
    "*": -1
}

```

## Load event data

```python
events = pd.read_csv("../data/events.csv")
 

events["dow_human"] = events["dow"] 
events["dow"] = events["dow"].replace(dow)
events
```

## Generate a blank schedule 

```python


start_date = '2025-08-25'  # Full start date
end_date = '2025-12-15'     # Full end date
num_rows = 5                # Number of rows

result_df = create_empty_sched_df(start_date, end_date, num_rows)
result_df
```
