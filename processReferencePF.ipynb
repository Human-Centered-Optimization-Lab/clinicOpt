# Process Reference PF
This script reads in the raw outputs of a list of outputs from the DESDEO RPM project and outputs a clean CSV of the PF

## Load variable

```python
import pickle
import numpy as np
import pandas as pd

output = open('pf_1295.pkl', 'rb')
raw_output = pickle.load(output)
```

## Parse

```python
output_flat = np.array(raw_output).flatten()


objectives = pd.DataFrame([output.optimal_objectives for output in output_flat])

objectives.drop_duplicates()

objectives.to_csv("pf_1295.csv")

```
